```python
# choose data
var_in = 'PRECT'
# Load data for larger area than only training area - set file name
fin_training = var_in+'_autumn_node_'+village_node+'_alaska_cam_grid'

# load data
ds = xr.open_dataset(dir_in+fin_training+'.nc')

# explicitly load data so that it doesn't take forever later on during mean
ds = ds.load()
```

## Break Data into decades

```python
# choose subset of decades to do initial analysis
decades = [1950,2000,2050]
decades_str = ['1950','2000','2050']
```

```python
# data for each decade in subset of decades
mask_1950 = ds.time.where(ds.decades == 1950)
prect_1950 = prect.where(mask_1950 >= 0, drop=True)
print('number of times in 1950:'+str(len(prect_1950.time)))

mask_2000 = ds.time.where(ds.decades == 2000)
prect_2000 = prect.where(mask_2000 >= 0, drop=True)
print('number of times in 2000:'+str(len(prect_2000.time)))

mask_2050 = ds.time.where(ds.decades == 2050)
prect_2050 = prect.where(mask_2050 >= 0, drop=True)
print('number of times in 2050:'+str(len(prect_2050.time)))
```

```python
units = ds.units
print(units)
```

## Get decadal composites and differences

```python
# average over all times in each decadal dataset
prect_1950_composite = prect_1950.mean(dim='time')
prect_2000_composite = prect_2000.mean(dim='time')
prect_2050_composite = prect_2050.mean(dim='time')
```

```python
# calculate differences from 1950
diff_2000 = prect_2000_composite - prect_1950_composite
diff_2050 = prect_2050_composite - prect_1950_composite
```

```python
# calculate percent differences from 1950
diff_pct_2000 = 100*((prect_2000_composite - prect_1950_composite)/((prect_2000_composite + prect_1950_composite)/2))
diff_pct_2050 = 100*((prect_2050_composite - prect_1950_composite)/((prect_2050_composite + prect_1950_composite)/2))
```
