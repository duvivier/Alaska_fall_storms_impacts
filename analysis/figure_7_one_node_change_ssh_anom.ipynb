### Get grid data

```python
# load grid files for plotting

dir_grid = '/glade/u/home/duvivier/masks/'
file_in = 'POP_tx0.1v3_grid.nc'
ds_grid = xr.open_mfdataset(dir_grid+file_in, decode_times=False)    

# clip global data to match other files - do in two steps because that's how I originally clipped data
# Step 1: clip to full Arctic Region
ds_grid_sub = ds_grid.isel(nlat=slice(1700,2250)).isel(nlon=slice(2700,3300))
```

## Break Data into decades

```python
# choose subset of decades to do initial analysis
decades = [1950,2000,2050]
decades_str = ['1950','2000','2050']
```

```python
# data for each decade in subset of decades
mask_1950 = ds.time.where(ds.decades == 1950)
ssh_anom_1950 = ssh_anom.where(mask_1950 >= 0, drop=True)
print('number of times in 1950:'+str(len(ssh_anom_1950.time)))

mask_2000 = ds.time.where(ds.decades == 2000)
ssh_anom_2000 = ssh_anom.where(mask_2000 >= 0, drop=True)
print('number of times in 2000:'+str(len(ssh_anom_2000.time)))

mask_2050 = ds.time.where(ds.decades == 2050)
ssh_anom_2050 = ssh_anom.where(mask_2050 >= 0, drop=True)
print('number of times in 2050:'+str(len(ssh_anom_2050.time)))
```

```python
units = ds.units
print(units)
```

## Get decadal composites and differences

```python
# average over all times in each decadal dataset
ssh_1950_composite = ssh_anom_1950.mean(dim='time')
ssh_2000_composite = ssh_anom_2000.mean(dim='time')
ssh_2050_composite = ssh_anom_2050.mean(dim='time')
```

```python
# calculate differences from 1950
diff_2000 = ssh_2000_composite - ssh_1950_composite
diff_2050 = ssh_2050_composite - ssh_1950_composite
```
